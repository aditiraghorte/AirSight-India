```python
def build_features(df, keep_cities, lags=(1, 2, 3)):
    d = df[df["City"].isin(keep_cities)].sort_values(
        ["City", "Date"]
    ).copy()

    g = d.groupby("City")

    pollutants = [
        c for c in d.columns
        if c not in ["City", "Date", "AQI"]
    ]

    new = {}

    for l in lags:
        new[f"AQI_lag{l}"] = g["AQI"].shift(l)

    for p in pollutants:
        new[f"{p}_lag1"] = g[p].shift(1)

    new["AQI_roll7"] = g["AQI"].transform(
        lambda x: x.shift(1).rolling(7, min_periods=4).mean()
    )

    new["PM2.5_roll7"] = g["PM2.5"].transform(
        lambda x: x.shift(1).rolling(7, min_periods=4).mean()
    )

    d = pd.concat([d, pd.DataFrame(new, index=d.index)], axis=1)

    d["month"] = d["Date"].dt.month
    d["dayofweek"] = d["Date"].dt.day_of_week

    d = d.drop(columns=pollutants)
    d = d.dropna(subset=["AQI", "AQI_lag1"])

    d["City"] = d["City"].astype("category")

    return d.reset_index(drop=True)
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor, HistGradientBoostingRegressor
from sklearn.pipeline import make_pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.compose import ColumnTransformer
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

train = feat[feat["Date"] < "2020-01-01"]
test = feat[feat["Date"] >= "2020-01-01"]
print(len(train), len(test), test["City"].nunique())

features = [c for c in feat.columns if c not in ["AQI", "Date"]]
num_cols = [c for c in features if c != "City"]
X_train, y_train = train[features], train["AQI"]
X_test, y_test = test[features], test["AQI"]

prep = ColumnTransformer([
    ("num", make_pipeline(SimpleImputer(strategy="median"), StandardScaler()), num_cols),
    ("city", OneHotEncoder(handle_unknown="ignore"), ["City"]),
])

def evaluate(name, pred):
    return {"Model": name,
            "MAE": mean_absolute_error(y_test, pred),
            "RMSE": mean_squared_error(y_test, pred) ** 0.5,
            "R2": r2_score(y_test, pred)}

results = [evaluate("Baseline (persistence)", test["AQI_lag1"])]

models = {
    "Linear Regression": make_pipeline(prep, LinearRegression()),
    "Random Forest": make_pipeline(prep, RandomForestRegressor(
        n_estimators=200, min_samples_leaf=3, n_jobs=-1, random_state=42)),
    "Gradient Boosting": HistGradientBoostingRegressor(
        categorical_features="from_dtype", random_state=42),
}

for name, m in models.items():
    m.fit(X_train, y_train)
    results.append(evaluate(name, m.predict(X_test)))

pd.DataFrame(results).round(2)
```

```python
bins = [0, 50, 100, 200, 300, 400, 500]
labels = ["Good", "Satisfactory", "Moderate", "Poor", "Very Poor", "Severe"]

feat["Bucket"] = pd.cut(
    feat["AQI"],
    bins=bins,
    labels=labels,
    include_lowest=True
)

feat["Bucket_lag1"] = pd.cut(
    feat["AQI_lag1"],
    bins=bins,
    labels=labels,
    include_lowest=True
)

ctest = feat[feat["Date"] >= "2020-01-01"]

print("Test class counts:")
print(ctest["Bucket"].value_counts())

print("\nMissing previous-day buckets:", ctest["Bucket_lag1"].isna().sum())
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, confusion_matrix, f1_score
from sklearn.pipeline import make_pipeline
import matplotlib.pyplot as plt
import seaborn as sns

c = feat.dropna(subset=["Bucket", "Bucket_lag1"]).copy()

ctrain = c[c["Date"] < "2020-01-01"]
ctest = c[c["Date"] >= "2020-01-01"]

clfs = {
    "Logistic Regression": make_pipeline(
        prep,
        LogisticRegression(max_iter=2000, class_weight="balanced")
    ),
    "Random Forest": make_pipeline(
        prep,
        RandomForestClassifier(
            n_estimators=200,
            min_samples_leaf=3,
            class_weight="balanced",
            n_jobs=-1,
            random_state=42
        )
    )
}

print(
    "Persistence macro F1:",
    round(f1_score(ctest["Bucket"], ctest["Bucket_lag1"], average="macro"), 3)
)

for name, model in clfs.items():
    model.fit(ctrain[features], ctrain["Bucket"].astype(str))
    pred = model.predict(ctest[features])

    print(name, "macro F1:",
          round(f1_score(ctest["Bucket"].astype(str), pred, average="macro"), 3))
```

```python
import os, joblib
from sklearn.model_selection import RandomizedSearchCV, TimeSeriesSplit
from sklearn.inspection import permutation_importance

train_s = train.sort_values("Date")
X_tr, y_tr = train_s[features], train_s["AQI"]

param_dist = {
    "learning_rate": [0.03, 0.05, 0.1],
    "max_iter": [200, 400, 600],
    "max_depth": [3, 5, None],
    "min_samples_leaf": [20, 50, 100],
    "l2_regularization": [0, 1, 5],
}
search = RandomizedSearchCV(
    HistGradientBoostingRegressor(categorical_features="from_dtype", random_state=42),
    param_dist, n_iter=15, cv=TimeSeriesSplit(n_splits=4),
    scoring="neg_mean_absolute_error", n_jobs=-1, random_state=42)
search.fit(X_tr, y_tr)

best = search.best_estimator_
pred = best.predict(X_test)
print(search.best_params_)
print("Tuned MAE:", round(mean_absolute_error(y_test, pred), 2),
      "| RMSE:", round(mean_squared_error(y_test, pred) ** 0.5, 2))

# what drives the model
imp = permutation_importance(best, X_test, y_test, n_repeats=5, random_state=42,
                             scoring="neg_mean_absolute_error")
print(pd.Series(imp.importances_mean, index=features).sort_values(ascending=False).head(10).round(2))

# where it fails
err = test.assign(pred=pred, abs_err=(test["AQI"] - pred).abs())
print(err.groupby("City")["abs_err"].mean().sort_values(ascending=False).round(1))
print(err.groupby(err["Date"].dt.month)["abs_err"].mean().round(1))

os.makedirs("../models", exist_ok=True)
joblib.dump(best, "../models/aqi_model.joblib")
```
